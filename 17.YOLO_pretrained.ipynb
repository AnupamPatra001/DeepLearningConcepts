```python
from ultralytics import YOLO


model = YOLO("yolo11m-cls.pt")  # load a pretrained model (recommended for training)
```

```python
# Predict with the model
results = model("https://ultralytics.com/images/bus.jpg")
```

```python
import requests

def download_image_from_url(url, save_path):
  """
  Downloads an image from a given URL and saves it to a local path.

  Args:
    url (str): The URL of the image.
    save_path (str): The local path to save the image to.
  """
  try:
    response = requests.get(url, stream=True)
    response.raise_for_status()  # Raise an exception for bad status codes

    with open(save_path, 'wb') as out_file:
      out_file.write(response.content)
    print(f"Image downloaded successfully to {save_path}")
  except requests.exceptions.RequestException as e:
    print(f"Error downloading image: {e}")
```

```python
# Example usage
download_image_from_url("https://i.redd.it/mpwy0owfoqp41.jpg", "vine_snake.jpg")
```

```python
display(Image(filename='vine_snake.jpg'))
```

```python
# Predict with the model
results = model("/content/vine_snake.jpg")
```
