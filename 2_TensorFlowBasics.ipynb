### Important TensorFlow functions

````python
# 1. tf.constant()

# Creates a tensor whose value is not intended to be changed.

x = tf.constant([1, 2, 3])


print(x)
````

````python
# 2. tf.Variable()
# Creates a mutable tensor.
# This is particularly important for model parameters such as weights and biases.

x = tf.Variable([1.0, 2.0, 3.0])
x.assign([4.0, 5.0, 6.0])
print(x)
````

````python
# 3. tf.zeros()
# Creates a tensor filled with zeros.

x = tf.zeros((2, 3))
print(x)
````

````python
# 4. tf.ones()
# Creates a tensor filled with ones.

x = tf.ones((2, 3))
print(x)
````

````python
# 5. tf.random
# Used for generating random tensors.
# For example:

x = tf.random.normal((2, 3))
print(x)

# This generates values approximately following a normal distribution.

# Very useful for:

# weight initialization,
# random data,
# experimentation
````

````python
# 6. tf.reshape()
# Changes the shape of a tensor without changing its underlying values

x = tf.constant([1, 2, 3, 4, 5, 6])
y = tf.reshape(x, (2, 3))

print(y)
````

````python
# 7. tf.cast()
# Changes the data type.

x = tf.constant([1, 2, 3])
y = tf.cast(x, tf.float32)

print(y)
````

````python
# 8. tf.concat()

# Joins tensors together.

a = tf.constant([[1, 2]])
b = tf.constant([[3, 4]])
c = tf.concat([a, b], axis=0)
d = tf.concat([a, b], axis=1)

print(c)
print(d)
````

````python
# 9. tf.stack()

# Stacks tensors along a new dimension.

a = tf.constant([1, 2])
b = tf.constant([3, 4])
c = tf.stack([a, b])

print(c)

# The important difference is that stack() introduces a new dimension, while concat() joins along an existing dimension.
````

````python
# 10.Mathematical operations
# TensorFlow supports mathematical operations directly.

a = tf.constant([1, 2, 3])
b = tf.constant([4, 5, 6])


print(tf.add(a, b))
print(tf.subtract(a, b))
print(tf.multiply(a, b))
print(tf.divide(a, b))
````

````python
# 11. Matrix multiplication
# Very important for neural networks.

a = tf.constant([[1, 2],
                 [3, 4]])
b = tf.constant([[5, 6],
                 [7, 8]])

c = tf.matmul(a, b)

print(c)
````

````python
# 12. Scalar multiplication
a = tf.constant([[1, 2],
                 [3, 4]])
b=4
c = tf.multiply(a, b)
print(c)
````

````python
# 13. tf.reduce_mean()
# Calculates the mean.

x = tf.constant([10.0, 20.0, 30.0])

print(tf.reduce_mean(x))
````

````python
# 14. sum, minimum, maximum, mean of array

x = tf.constant([1, 2, 3])

print(tf.reduce_sum(x))
print(tf.reduce_min(x))
print(tf.reduce_max(x))
print(tf.reduce_mean(x))
````

````python
# 15. sum, minimum, maximum, mean of 2darray

x = tf.constant([[1, 2, 3],[4,5,6]])

print(tf.reduce_sum(x))
print(tf.reduce_min(x))
print(tf.reduce_max(x))
print(tf.reduce_mean(x))
````

````python
#  16. operating tensor with scalar
raw=tf.constant([[2,4,6],[8,10,12],[14,6,8]])

print(raw + 2)
print(raw - 2)
print(raw * 2)
print(raw / 2)
````

````python
#  17. Dot product
a = tf.constant([[1, 2],
                 [3, 4]])
b = tf.constant([[5, 6],
                 [7, 8]])

print(tf.matmul(a, b))
print(tf.tensordot(a, b, axes=0))
print(tf.tensordot(a, b, axes=1))
print(tf.tensordot(b, a, axes=0))
print(tf.tensordot(b, a, axes=1))
````

### What is Keras?
- **Keras is a high-level deep learning API used to build and train neural networks.**

- Today, Keras is commonly used with TensorFlow.

You can think of it like:
```
TensorFlow = underlying deep learning framework
Keras = easier interface for building neural networks
```
For example, instead of manually implementing every operation, you can write:

````python
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax")
])
print(model)
````

**Why do we need Keras?**

- Suppose you want to build a neural network.

- Without a high-level API, you would need to manually handle:
```
tensors
weights
matrix multiplication
biases
activation functions
gradients
backpropagation
parameter updates
```
- Keras provides abstractions for these.

- So instead of implementing everything manually all these processes, you can simply specify:
```
model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax")
])
```
- Keras handles much of the underlying computation.

### What is a Neural Network?

**A neural network is a machine learning model composed of interconnected layers of artificial neurons that learn patterns from data.**

- A basic neural network looks like:
```
Input Layer
     ↓
Hidden Layer
     ↓
Hidden Layer
     ↓
Output Layer
```
* For example, suppose we're predicting whether an email is spam.
```
Email Features
     ↓
Input Layer
     ↓
Hidden Layer
     ↓
Hidden Layer
     ↓
Output Layer
     ↓
Spam / Not Spam
```

### What is a neuron?

**A neuron takes inputs, applies weights and bias, and then passes the result through an activation function.**

Mathematically:



Then:
```
output=f(z)
```

where:
```
x = input
w = weight
b = bias
z = weighted sum
f = activation function
```
- This is the fundamental operation of a neural network.

**What does Dense mean?**

- Dense means fully connected layer.
- Every neuron in one Dense layer is connected to every output from the previous layer.

### What is an activation function?

- This is extremely important.
- An activation function determines the output of a neuron after the weighted sum.
- Without activation functions, stacking multiple neural network layers would still result in an overall linear transformation.
- Activation functions introduce non-linearity, allowing neural networks to learn complex relationships.

The general structure is:
```
z= Wx+b
```
then
```
a=f(z)
```
where f is activation function.

```
- TensorFlow → framework for computation and deep learning.
- Keras → high-level API for building neural networks.
- Tensor → data structure used to represent numerical data.
- Neuron → computes weighted inputs + bias and applies an activation function.
- Layer → collection/transformation of neurons.
- Activation function → introduces non-linearity.
- Forward propagation → produces prediction.
- Loss function → measures prediction error.
- Backpropagation → calculates gradients.
- Optimizer → uses gradients to update weights.
- Epoch → one complete pass through training data.
- Batch → subset of training data processed in one iteration.
```

**Tensors**
A tensor is a container for numerical data arranged in $n$ dimensions. In machine learning, tensors represent inputs, weights, and intermediate activations.

* **Rank 0 (Scalar):** Single number (e.g., `5`, shape `()`).
* **Rank 1 (Vector):** 1D array representing a list of features (e.g., `[1, 2, 3]`, shape `(3,)`).
* **Rank 2 (Matrix):** 2D table representing tabular data or batches of vectors (e.g., shape `(batch_size, features)`).
* **Rank 3 (3D Tensor):** Sequence or time-series data (e.g., shape `(batch_size, timesteps, features)`).
* **Rank 4 (4D Tensor):** Image batches (e.g., shape `(batch_size, height, width, channels)`).

---

**Weights**
Weights are learnable parameters that determine the strength and direction of the connection between an input feature and a downstream neuron.

* **Kernel / Dense Weights:** 2D matrices connecting inputs to neurons in fully connected layers.
* **Convolutional Filters (Kernels):** Small sliding matrices (e.g., $3 \times 3$) used to capture spatial patterns in computer vision.
* **Recurrent Weights:** Matrices that transform hidden states across time steps in sequential models.

---

**Matrix Multiplication**
Matrix multiplication computes linear combinations of all inputs for all neurons simultaneously: $Z = XW$. If input batch $X$ has shape $(N, D_{in})$ and weight matrix $W$ has shape $(D_{in}, D_{out})$, the dot product projects the inputs into the next layer’s feature space of shape $(N, D_{out})$.

* **Standard Dot Product ($XW$):** Linear transformations in Dense layers.
* **Batched MatMul:** Parallel matrix multiplications across multi-head attention blocks in Transformers.
* **Element-wise Multiplication (Hadamard product):** Gating mechanisms (e.g., in LSTMs or GRUs), distinct from true matrix multiplication.

---

**Biases**
Biases are learnable additive constants added to the weighted sum: $Z = XW + b$. They allow the activation function to shift horizontally (left or right), enabling neurons to activate even when all inputs are zero.

* **Per-Neuron Bias:** A 1D vector of length $D_{out}$ broadcasted across every row in a batch.
* **Per-Channel Bias:** A single bias term applied across each spatial channel in Convolutional layers.

---

**Activation Functions**
Activation functions apply element-wise non-linear mappings $A = f(Z)$. Without them, stacking multiple layers collapses mathematically into a single linear transformation.

* **ReLU (Rectified Linear Unit):** $f(z) = \max(0, z)$. Fast to compute and mitigates vanishing gradients for positive inputs; standard for hidden layers.
* **Leaky ReLU / GELU:** Fix dead neurons by allowing a small slope for $z < 0$. GELU is standard in modern Transformer architectures.
* **Sigmoid:** $\sigma(z) = \frac{1}{1 + e^{-z}}$. Maps outputs to $(0, 1)$; used for binary classification outputs.
* **Softmax:** Multi-class generalization of sigmoid; normalizes a vector of logits into a probability distribution summing to 1.
* **Tanh:** Maps outputs to $(-1, 1)$; zero-centered, commonly used in RNNs.

---

**Gradients**
A gradient is a vector of partial derivatives ($\nabla_W L$) representing the rate of change of the scalar loss $L$ with respect to each model parameter. It points in the direction of the steepest ascent of the error surface; moving opposite to the gradient reduces the loss.

* **Weight Gradients ($\frac{\partial L}{\partial W}$):** Direct signal indicating how to nudge weights.
* **Input Gradients ($\frac{\partial L}{\partial X}$):** Used for adversarial attacks, feature visualization, and passing errors back to preceding layers.

---

**Backpropagation**
Backpropagation is the efficient application of the calculus Chain Rule to calculate gradients from the output layer back to the input layer.

* **Forward Pass:** Compute linear combinations and activations layer-by-layer to generate predictions and calculate the loss.
* **Backward Pass:** Propagate error backward by multiplying local derivatives (the gradient of the activation function with respect to its input) by the downstream gradient from higher layers.

---

**Parameter Updates**
Once gradients are calculated, an optimization algorithm adjusts the parameters to minimize the loss function: $W_{new} = W_{old} - \eta \cdot \Delta W$, where $\eta$ is the learning rate.

* **SGD (Stochastic Gradient Descent):** Simple step proportional to the negative gradient; often supplemented with Momentum.
* **RMSprop:** Adapts the learning rate per parameter by dividing by a running average of recent gradient magnitudes.
* **Adam / AdamW:** Combines Momentum (first moment) with adaptive scaling (second moment); AdamW decouples weight decay and is the default optimizer for deep learning.
