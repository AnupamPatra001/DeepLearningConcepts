Imports

```python
from tensorflow import keras
from keras.datasets import mnist
from keras.layers import MaxPool2D, AvgPool2D, Conv2D, Flatten, Dense
from keras.models import Sequential
```

Performing dataload and reshaping

```python
(x_train, y_train), (x_test, y_test)=mnist.load_data()
print(x_train.shape)
# reshaping
x_train=x_train.reshape(x_train.shape[0], 28, 28, 1)
x_test=x_test.reshape(x_test.shape[0], 28, 28, 1)
```

```python
x_train.shape
```

```python
y_train
```

Normalization

```python
x_train=x_train / 255
x_test=x_test / 255
```

```python
x_test[1]
```

One hot encoding

```python
y_train = keras.utils.to_categorical(y_train, 10)
y_test = keras.utils.to_categorical(y_test, 10)
```

Creating a LeNet5 architecture according to research paper

```python
model=Sequential()
model.add(Conv2D(filters=6, kernel_size=(5, 5), activation='relu', input_shape=(28, 28, 1)))  #(24, 24, 6)
model.add(MaxPool2D(pool_size=(2,2))) #(12, 12, 6)
model.add(Conv2D(filters=16, kernel_size=(5, 5), activation='relu'))  #(8, 8, 16)
model.add(MaxPool2D(pool_size=(2,2))) #(4, 4, 16)
model.add(Conv2D(filters=120, kernel_size=(5, 5), activation='relu', padding='same'))  #(4, 4, 16)
model.add(Flatten())
model.add(Dense(84, activation='relu'))
model.add(Dense(10, activation='softmax'))
```

```python
model.summary()
```

Modelcompilation

```python
model.compile(
    optimizer=keras.optimizers.Adam(),
    loss=keras.losses.CategoricalCrossentropy(),
    metrics=['accuracy']
)
model.fit(x_train, y_train, batch_size=128, epochs=10, verbose=1, validation_data=(x_test, y_test))
```

```python
train_score=model.evaluate(x_train, y_train)
test_score=model.evaluate(x_test, y_test)
print('train_loss:', train_score[0])
print('test_loss:', test_score[0])
print('train_acc:', train_score[1])
print('test_acc:', test_score[1])
```
